```python
# Import the necessary packages
import pickle
import numpy as np
import cv2
```

```python
def preprocess_image(img_path):
    image = cv2.imread(img_path)
    if image is None:
        raise Exception("\nImage not read successfully.\n")
    height, width, channels = image.shape
    if width<height :
        image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    target_size = (280, 180)
    preprocessed_image = cv2.resize(image, target_size)
    return preprocessed_image

def extract_saturation_hist(bgr_image):
    # use only saturation channel images
    converted_image = cv2.cvtColor(bgr_image, cv2.COLOR_BGR2HSV)
    s_channel = converted_image[:,:,1]
    s_hist = cv2.calcHist([s_channel], [0], None, [256], [0, 256])
    s_hist /= s_hist.sum()
    s_hist = s_hist.squeeze()
    return s_hist

def load_model(model_path):
    with open(model_path, 'rb') as f:
        model = pickle.load(f)
    return model

def predict(model_path, image_path):
    label_map = {0: 'Bad', 1: 'Good'}
    image = preprocess_image(image_path)
    s_hist = extract_saturation_hist(image)
    s_hist = s_hist.reshape(1, *s_hist.shape)
    model = load_model(model_path)
    pred = model.predict(s_hist)
    predicted_class = label_map[pred[0]]
    return predicted_class
```

```python
model_filepath = r"/content/drive/MyDrive/Sowmya /models/Sardine/v1/svm77_saturation_hist.pkl"
img_path = r"/content/drive/MyDrive/Sowmya /qZense Dataset/Final Data/Sardine/Bad/20230523094036995_sardine_bad.jpeg"
predicted_class = predict(model_filepath, img_path)
print(f"Predicted class: {predicted_class}")
```
