this yolo segmentation model for tomato works only when ultralytics version is 8.3.1 and python 3.10.14

```python
import ultralytics
print(ultralytics.__version__)
```

```python
from ultralytics import YOLO

model = YOLO(r"C:\Users\sowmy\Downloads\tomato segmentation\train\weights\best.pt")
```

```python
test_folder = r"C:\Users\sowmy\Downloads\tomato segmentation\unseen"
results = model(test_folder)
import matplotlib.pyplot as plt
for i in range(len(results)):
    plt.figure(figsize=(10, 10))
    plt.imshow(results[i].plot())
    plt.show()
```
