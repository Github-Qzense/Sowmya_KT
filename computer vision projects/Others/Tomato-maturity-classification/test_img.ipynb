```python
from tensorflow.keras.models import load_model
loaded_model = load_model("/content/drive/MyDrive/Sowmya /models/experiments/Tomato classification/model.h5")
```

```python
from tensorflow.keras.applications.densenet import preprocess_input
from tensorflow.keras.preprocessing import image
import tensorflow as tf
import numpy as np

# Define image loading function
def load_image(image_path):
    image = tf.io.read_file(image_path)
    image = tf.io.decode_jpeg(image, channels=3)
    image = tf.image.resize(image, [224, 224])
    image = preprocess_input(image)
    return image

test_img = "/content/drive/MyDrive/Sowmya /models/experiments/Tomato classification/41av+vz+ppL.jpg"

test_img = load_image(test_img)

# Make predictions
p = loaded_model.predict(np.array([test_img])).flatten()
threshold = 0.5
# p = np.where(p > threshold, 1, 0)
if p[0] > threshold:
    print("Mature")
else:
    print("Raw")
```
